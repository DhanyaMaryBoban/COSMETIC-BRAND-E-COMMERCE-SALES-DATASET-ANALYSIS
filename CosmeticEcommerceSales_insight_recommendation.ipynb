# COSMETIC BRAND E-COMMERCE SALES DATASET ANALYSIS

## Step 4:  Insight Generation and Report 

##### **Goal**: To summarize key insights from analysis

### Introduction to Findings

This report presents a comprehensive analysis of the cosmetic brand’s e-commerce sales data, encompassing 8,165 transactions across 18 attributes. The goal was to uncover actionable insights across four key dimensions: Time-related, Transaction-related, Customer-related, and Product-related attributes. By classifying and comparing these insights, we reveal patterns of **consumer behavior, operational strengths and weaknesses, and strategic opportunities for growth**.

### Insight Framework: Time, Transaction, Customer, Product

#### 1. Time-related Attributes

- Sales Trend Over Time: **October** peak sales, sharp drop in November → strong seasonality.
- Correlation with Time Factor: Weak correlation between time and gross price → **prices remain stable** across calendar periods.

#### 2. Transaction-related Attributes

- Order Status Analysis: **5,788 delivered orders (high satisfaction)**, 2,329 in transit (logistics bottleneck).
- Gross Amount by Order Status: **Delivered orders dominate revenue**; pending statuses contribute minimally.
- Correlation (Gross Amount ↔ Order Quantity): Moderate positive correlation → customers **not highly price-sensitive**.

#### 3. Customer-related Attributes

- Unique vs. Repeat Customers: **76%** repeat purchase rate → strong loyalty; 487 one-time buyers → conversion opportunity.
- Regional Sales Performance (City & Zone): **Datia** dominates; **West** Zone leads in average sales.
- Sales by State: Maharashtra highest demand; West Bengal & Delhi strong secondary markets.

#### 4. Product-related Attributes

- Average Order Value by Product: **Premium driver (Insight Glitter Purse**) at ₹800; core cosmetics cluster at ₹215–₹300.
- Top Products: **Mini Power Matte Lipcolour** most repeated product.
- Popular Variants & Skin Tone Preferences: **Prime It Up** most popular across all tones; Coral Dream & Divine Wine strong secondary variants.


### Highlighted Significant Patterns, Correlations, and Anomalies

#### Patterns:

 - Strong customer loyalty (76% repeat buyers).
 - Seasonal demand peak in October.
 - Premium products sell well despite higher prices.
 - Datia and West Zone are key revenue drivers.

#### Correlations:

- Moderate positive correlation between Gross Amount and Order Quantity → higher-priced products still sell well.
- Weak correlation between Time and Gross Price → product mix matters more than timing.

#### Anomalies:

- High number of in-transit orders (2,329) → logistics bottleneck (indicator of process inefficiency).
- Sharp drop in November sales → reliance on festive campaigns.
- Heavy geographic imbalance with Datia dominating sales.

### Overall Summary & Recommendations

The dataset reveals a loyal customer base, strong seasonal peaks, and regional concentration in Datia and West Zone. However, anomalies such as logistics delays, sharp post-festive drops, and geographic imbalance highlight areas needing improvement.

#### Next Steps:

- Launch **loyalty programs and personalized offers** to encourage loyal buyers.
- Strengthen **courier partnerships and real-time tracking** to reduce transit delays.
- **Expand beyond Datia and strengthen secondary cities/zones** to reduce geographic dependency.
-  **Build campaigns around hero products** (Prime It Ups, Mini Power Matte Lipcolour) to lift AOV.
-  Design **festive and off season promotional calendars** to sustain demand across cycle.
-  Prioritize **inventory establishment in Maharashtra**, while **strengthening campaigns** in West Bengal, Delhi, Karnataka, and Uttar Pradesh to boost demand.
-  Since higher‑priced products are still selling well, selectively **increase prices on high‑demand products and variants to improve margins without hurting sales**.For example: Since Prime It Up sells strongly even at ₹406, a small price increase to ₹410–₹415 could boost margins without reducing demand. 

### Conclusion:

This e‑commerce sales analysis not only identifies key performance trends but also clarifies the purpose of the study—to transform raw data into actionable insights. By highlighting consumer behavior, operational challenges, and growth opportunities, the analysis serves as a decision‑making tool that enables businesses to adapt strategies, strengthen customer relationships, and drive sustainable success in the digital marketplace.
