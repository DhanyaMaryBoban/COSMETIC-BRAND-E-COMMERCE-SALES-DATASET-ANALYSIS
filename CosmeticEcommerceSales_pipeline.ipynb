### 2.4 Derived Columns and Aggregated Metrics in Analysis

#### Calculate Sales

````python
# Sales = Gross Amount * Order Quantity

df['Sales'] = df['Gross amount'] * df['Order quantity']
````

````python
df.head()
````

### Compute Average Order Value (AOV)

````python
# Step 1: Calculate Total sales 
Total_Sales = df['Sales'].sum()
print("Total Sales:", Total_Sales)

# Step 2: Count Total Orders
Total_Orders = df["Order Id"].nunique()
print("Total Orders:", Total_Orders)

#Step 3: Compute AOV by dividing total sales by total orders
AOV = Total_Sales / Total_Orders 
print("Average Order Value:", AOV)
````

##### Insight from these metrics:
- The business generated over 1 million in revenue from 2,217 orders. That’s a healthy topline figure.
- On average, each order contributes about ₹477. This tells us the typical customer spend per transaction.

````python
#Import Visualization Libraries

import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

# Display setting for Visualization

plt.style.use("default")
sns.set_context("notebook")
````

### 1. Unique Customers and Repeat Purchase Rate

````python
# Display Unique customers

unique_customers = df['Customer Code'].nunique()
print("Unique customers:",unique_customers)

# Repeat purchase rate

repeat_customers = df['Customer Code'].value_counts()[df['Customer Code'].value_counts()>1].count()
print("Repeat customers:", repeat_customers)
````

##### Insight & Recommendation:
- Unique customers: 2012 placed at least one order 
- Repeat customers: 1525 (76% of total) → strong loyalty, major strength
- One-time customers: 2012-1525 = 487
- Focus on converting one‑time buyers into repeat customers while nurturing the loyal base.


### 2. Order Status Analysis

````python
# Order status analysis

df['Order Status'].value_counts()
````

##### Insight & Recommendation:
- 5,788 orders successfully delivered, reflecting high customer satisfaction.
- 2,329 orders still in transit — requires monitoring of delivery times and courier performance.
- Only 32 booked and 15 dispatched, showing minimal early-stage delays
- Provide real‑time tracking and proactive updates to reassure customers about shipped orders.

### 3. AOV by Product

````python
# Display AOV by product

aov_by_product = df.groupby('Product Name')['Sales'].mean().sort_values(ascending=False)
print(aov_by_product.head(10))
````

##### Insight and Recommendation:
- Premium driver: Insight Glitter Purse has the highest AOV (₹800) → strong revenue contributor despite likely lower sales volume.
- Core cosmetics: Most face, lips, and eyes products cluster around ₹215–₹300 AOV → affordable, everyday items driving volume but not value.
- Opportunity: Bundling and cross‑selling cosmetics can lift their AOV, reducing reliance on premium items alone.

### 4. Sales by City

````python
# Display Sales by City
sales_by_city = df.groupby('City')['Sales'].mean().sort_values(ascending=False)
print(sales_by_city.head(5))
````

##### Insight and Recommendation:
- Sales are heavily concentrated in Datia.
- Other cities show low to moderate contributions.
- Strategy should focus on protecting Datia’s demand.
-  Adopt strategory to boost sales in secondary cities to reduce dependency and spread growth.


### 5. Sales by Zone

````python
# Display Sales by Zone

sales_by_zone = df.groupby('Zone')['Sales'].mean().sort_values(ascending=False)
print(sales_by_zone.head())
````

##### Insight and Recommendation:
- West Zone leads with the highest average sales (₹135.69), showing stronger customer spend per order.
- South, North, and East Zones are close behind (₹124–₹131), indicating balanced but slightly lower in sales.
- Strategy should focus on leveraging West’s strength while uplifting other zones through targeted promotions or bundling.

### 6. Sales Trend over Time using Line Chart

````python
#  Sales trend over time 

df.groupby('Week Ending Date')['Sales'].sum().plot(kind='line') # creates a line chart of those weekly totals
plt.title('Weekly Sales Trend') 
plt.ylabel('Sales') 
plt.xlabel('Week Ending Date') 
plt.show()
````

##### Insight and Recommendation:
- October is the sales peak month, mostly driven by festive season or major campaigns..
- Sharp drop in November.
- Boost sales in peak season.
- Use promotions to lift off‑season demand.

### 7. Gross amount by Order Status using Box plot

````python
# Display Gross amount by Order Status 

sns.boxplot(x='Order Status', y='Sales', data=df)
plt.title('Gross Amount by Order Status') 
plt.show()
````

##### Insight and Recommendation:
- Delivered orders drive the bulk of revenue.
- Other statuses (shipped, dispatched, booked) contribute much smaller amounts.
- The boxplot makes it clear that converting pending statuses into delivery orders is the key to maximise revenue.

### 8. Top Products by State

````python
#Display Top Products by State

top_sale_products=df.groupby(['State','Product Name'])['Order quantity'].sum().sort_values(ascending=False)
print(top_sale_products.head())
````

##### Insight and Recommendation:

- Maharashtra: Highest demand → prioritize inventory and marketing.
- West Bengal & Delhi: Strong secondary markets → strengthen campaigns to boost demand.
- Karnataka & Uttar Pradesh: Moderate sales → use promotions to drive growth.
- Mini Power Matte Lipcolour: Most repeated product → Use its popularity to promote it across India.

### 9. Popular Product Variant and Skin Tones

````python
# Display Popular Product Variant and Skin Tones

popular_variant_skin_tones =df.groupby(['Variant Name','Skin Tones'])['Order quantity'].sum().sort_values(ascending=False)
print(popular_variant_skin_tones.head())
````

##### Insight and Recommendation:
- Prime It Up is the most popular variant across all skin tones.
- Coral Dream and Divine Wine follows it.
- Other variants also shows steady demand.
- All top variants are ordered across Fair, Medium, and Dusky tones, indicating  popular across multiple skin tones.
- Main focus on Prime It Up.
- Promote Coral Dream and Divine Wine as supporting options.

### 10. Correlation matrix using Heatmap

````python
# Display Correlation Matrix
corr = df[['Order quantity','Gross amount','Day','Month','Week']].corr() 
sns.heatmap(corr, annot=True, cmap='coolwarm') 
plt.title('Correlation Heatmap')
plt.show()
````

##### Insight and Recommendation:

- There is a moderate positive relationship between Gross amount and Order Quantity. This means customers are not price sensitive, here higher priced products are still selling well.
- Correlation between time factor and gross price is very weak. Price stay stable across calendar periods.
- Promote premium products.
- Market by product mix, not by time.
